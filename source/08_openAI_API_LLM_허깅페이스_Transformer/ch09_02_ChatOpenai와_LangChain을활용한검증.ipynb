# 2. LLM 답변 생성
## 2.1 openAI SDK를 선택
- 세밀한 제어가 필요할 때, 성능 최적화가 중요할 때

```python
from openai import OpenAI
client = OpenAI()
response = client.chat.completions.create(
    model = 'gpt-4.1-nano',
    messages = [
      {'role':"system","content":"당신은 친절하게 대답하는 비서입니다."},
      {'role' : "user", "content":"2020년 월드시리즈 우승자는 누구인가요?"}  
    ],
)
print(response)
```

```python
response.choices[0].message.content
```

## 2.2 LangChain을 선택
- 내부적으로 자동 메세지 형태를 변환해주고자 할 때

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model = 'gpt-4.1-nano', # 기본 모델 gpt-3.5-turbo
)
ai_message = llm.invoke("2020년 월드시리지 누가 우승했나요?")
```
