```python
import tensorflow as tf
from tensorflow.keras.models import Sequential # 모델 객체 생성
from tensorflow.keras.layers import Dense, Input # Layer 층을 쌓을 때 필요
import numpy as np
import pandas as pd
```

```python
# 1. data set
# 2. data 전처리
X_data = np.array([[0,0],
                  [0,1],
                  [1,0], 
                  [1,1]])
y_data = np.array([[0],[1],[1],[0]])

# 3. 모델 구성
model = Sequential()
# model.add(Input(shape=(2,))) # 입력층
model.add(Dense(units =10, # 출력
                input_dim=2, # 입력 : 입력층이 없으면 사용
                activation='relu')) # 활성화 함수
model.add(Dense(units=20, activation='relu'))
model.add(Dense(units=10, activation='relu'))
model.add(Dense(units=1, activation='sigmoid')) # 출력층, 이진분류에서 활성화 함수는 sigmoid
# 만약 다중분류라면 activation = 'softmax'(출력합을 1로)
model.summary()
```

```python
# 4. 모델 학습 과정 설정
model.compile(loss='binary_crossentropy',
              optimizer = 'adam',
              metrics=['binary_accuracy']) # 2진분류
```

```python
# 5. 학습
hist = model.fit(X_data, y_data, # 훈련데이터
                epochs = 600, # 학습 회수
                verbose = 2, # 0 : 학습로그 출력 X, 1: 학습로그 간소화 출력, 2: 학습로그 상세 출력
                )
```

```python
# 5. 학습 과정 살펴보기
hist.history.keys() # loss, binary_accuracy => compile 단계에서 추가하지 않으면 loss만 잡혀있음
import matplotlib.pyplot as plt
fig, loss_ax = plt.subplots(figsize=(10,5))
loss_ax.plot(hist.history.get('loss'), label = 'loss',color='red')
loss_ax.set_xlabel('epochs')
loss_ax.set_ylabel('loss')
accu_ax = loss_ax.twinx()
accu_ax.plot(hist.history.get('binary_accuracy'),label='accuracy',color='blue')
accu_ax.set_ylabel('accuracy')
loss_ax.legend(loc=(0.8,0.7))
accu_ax.legend(loc=(0.8,0.8))
plt.show()
```

```python
# 6. 학습 평가하기
score = model.evaluate(X_data,y_data, verbose=2) # loss, binary_accuracy
print(f'loss : {score[0]}')
print(f'binary_accuracy : {score[1]}')
```

```python
# 7. 모델 사용
model.predict(np.array([[5,1],[0,1]])).round()
```
