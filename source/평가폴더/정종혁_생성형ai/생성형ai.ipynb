```python
# 필요 라이브러리 import
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
```

# 1. LLM 모델 생성

```python
llm = ChatOllama(model='llama3.2:1b')
# 모델 체크
result = llm.invoke('plz tell me your information, Return format name : your name, feature : your feature')
result.content
```

# 2. 랭체인 생성
## 2-1. 국가 이름 => 대표 음식 return

```python
prompt_food_return = PromptTemplate(template="What is the representative dish of that {country}? you must Return the name of dish only.", input_variables=['country'])
output_parser = StrOutputParser()
food_chain = prompt_food_return | llm | output_parser
food_chain.invoke({'country':'Korea'})
```

## 2-2. 대표음식 -> 레시피 return

```python
prompt_recipe_return = PromptTemplate(template="""How to make {dish}? You must return whit it format like :
                                <How to cook the name of dish>
                               1. Preparing Ingredint
                               2. Ingredient prep
                               3. how to cook ... etc""", input_variables=['dish'])
recipe_chain = prompt_recipe_return | llm | output_parser
recipe_chain.invoke('bibimbab')
```

# 3. 랭체인 연결 및 실행

```python
final_chain = {'country':RunnablePassthrough()}|{'dish':food_chain}|recipe_chain
print(final_chain.invoke('Korea'))
```
