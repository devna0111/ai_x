```python
def fillna_nan(Serires) :
    Serires = Serires.copy()
    if pd.isnull(Serires['분양가격(제곱미터)']) :
        Serires['분양가격(제곱미터)'] = np.nan
    elif isinstance(Serires['분양가격(제곱미터)'], str) :
        if Serires['분양가격(제곱미터)'].replace(',','').replace(' ','') == '':
            Serires['분양가격(제곱미터)'] = np.nan
        elif Serires['분양가격(제곱미터)'].replace(',','').replace(' ',''):
            Serires['분양가격(제곱미터)'] = int(Serires['분양가격(제곱미터)'].replace(',','').replace(' ',''))
    return Serires
```

```python
fillna_nan(df_last.iloc[368])
```

```python
df_last = df_last.apply(fillna_nan,axis = 1)
```

```python
df_last['분양가격(제곱미터)'].astype('float64')
```

```python
df_last['평당분양가격'] = df_last['분양가격(제곱미터)'].astype('float64') * 3.3
```

```python
df_last.drop('분양가격(제곱미터)', axis = 1, inplace=True)
```

```python
df_last['전용면적'] = df_last['규모구분'].str.replace('전용면적','').str.replace('제곱미터','㎡').str.replace('초과','~').str.replace('이하','').str.replace(' ','').str.strip()
```

```python
df_last
```

```python
df_last.drop('규모구분',axis=1,inplace=True)
df_last
```

```python
df_last = df_last[['지역명','전용면적','연도','월','평당분양가격']]
df_last
```

# 5. 데이터 합치기

```python
df_first.reset_index(inplace=True)
```

```python
df_first = df_first.melt(id_vars='지역',var_name = '연도월', value_name = '평당분양가격')
df_first
```

```python
df_first
```

```python
def split_date(row) :
    row = row.copy()
    row['연도'] = row['연도월'].split('년')[0]
    row['월'] = row['연도월'].split('년')[1][:-1]
    return row
```

```python
df_first = df_first.apply(split_date, axis = 1)
```

```python
df_first.drop('연도월',axis=1, inplace=True)
df_first
```

```python
df_last2 = df_last[df_last.전용면적 == '모든면적']
```

```python
df_last2
```

```python
df_last2 = df_last2.drop('전용면적', axis = 1)
df_last2
```

```python
df_first['지역명'] = df_first['지역']
```

```python
df_first.drop('지역',axis=1, inplace = True)
df_first
```

```python
df_answer = pd.concat([df_first,df_last2],axis=0)[['지역명', '연도', '월', '평당분양가격']]
```

```python
df_answer = df_answer.reset_index(drop=True)
```
